```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
engine = create_engine('postgresql://postgres:1234@localhost:5432/netflix_db')
```

```python
df = pd.read_sql('SELECT * FROM tbl_netflix', engine)
df.head()
```

```python
print(df.shape)
print(df.dtypes)
print(df.describe())
```

```python
df['date_added'] = pd.to_datetime(df['date_added'])
print(df.dtypes)
```

```python
# quantos filmes e séries tem no catálogo?
df.groupby(['type'])['title'].count()
```

```python
#  quais os 10 países que mais produzem conteúdo? 
df['country'].value_counts(ascending=False).head(10)
```

```python
#  quantos títulos foram adicionados por ano?
df['date_added'].dt.year.value_counts().sort_index()
```

```python
# quais são os gêneros mais comuns?
df['listed_in'].value_counts(ascending=False).head(20)
```
